```python
str1 = "Hello"
Name = input("Enter Your Name:")
Final = str1+" "+Name
print(Final)
```

```python
str2 = ", Welcome to Python Programming"
Final_Str = Final+str2
print(Final_Str)
```

```python
a = Final_Str[0]
print("First Character:",a)
```

```python
b = Final_Str[-1]
print("Last Charcter:",b)
```

```python
c = Final_Str[0:5]
print("First 5 Characters:",c)
```

```python
d = Final_Str[-11:]
print("Last 11 Characters:",d)
```

```python
e = Final_Str[::-1]
print("Reversed String:",e)
```

```python
f = Final_Str[24:30]
print("Word Python:",f)
```

```python
strM = "Python beginner tutorial"
print(strM)
```

```python
Lower = strM.lower()
print("Lower Case:",Lower)
```

```python
Capital = strM.capitalize()
print("Capitalize:",Capital)
```

```python
occured = strM.count("t")
print("Occurence of t:",occured)
```
