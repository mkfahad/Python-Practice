```python
import numpy as np
import pandas as pd
```

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
print("Shape:",temperatures_w1.shape)
print("Datatype:",temperatures_w1.dtype)
print("Number of Elements:",temperatures_w1.size)
```

```python
temperatures_f = (temperatures_w1 * 9/5) + 32
print(temperatures_f)
```

```python
print("Max Temperature:",temperatures_w1.max())
print("Min Temperature:",temperatures_w1.min())
print("Mean Temperature:",round(temperatures_w1.mean(),2))
```

```python
print("Temperatures for First 3 days:",temperatures_w1[0:3])
print("Temperatures for the Weekend:",temperatures_w1[-2:])
print("Temperatures for the middle three days:",temperatures_w1[2:5])
```
